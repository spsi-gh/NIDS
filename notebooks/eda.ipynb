# UNSW-NB15 exploratory analysis
Run `python -m src.preprocess` first. Charts are also saved in `reports/eda_charts.html`.


```python
from pathlib import Path
import pandas as pd
import plotly.express as px
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
train = pd.read_csv(root / 'data/processed/UNSW_NB15_training-set.csv')
test = pd.read_csv(root / 'data/processed/UNSW_NB15_testing-set.csv')
print('train', train.shape, 'test', test.shape)
display(train.dtypes.to_frame('dtype'))
display(train.isna().sum().to_frame('missing'))
```

```python
display(train['label'].value_counts().rename(index={0:'Normal',1:'Attack'}))
display(train['attack_cat'].value_counts())
px.histogram(train, x='label', title='Normal vs attack').show()
px.bar(train.loc[train.label == 1, 'attack_cat'].value_counts(), title='Attack types').show()
```

```python
for column in ('proto', 'service'):
    px.bar(train[column].value_counts().head(12), title=f'Most common {column}').show()
px.histogram(train.sample(10000, random_state=42), x='dur', color='label', log_y=True, title='Flow duration').show()
numeric = ['dur','rate','spkts','dpkts','sbytes','dbytes','sttl','dttl','sload','dload']
px.imshow(train[numeric].corr(), color_continuous_scale='RdBu_r', zmin=-1, zmax=1, title='Correlations').show()
```
